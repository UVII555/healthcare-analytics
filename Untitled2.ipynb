```python
def print_opd_load_table(opd_data: dict):
    """Prints the OPD load dictionary as a clean Markdown table"""
    if not opd_data:
        print("No data available.")
        return

    # Header
    markdown = "| Appointment Date | Completed Appointments |\n"
    markdown += "| :--- | :---: |\n"
    
    # Rows (sorted chronologically or by most recent)
    for date, count in sorted(opd_data.items(), reverse=True):
        markdown += f"| {date} | {count} |\n"
        
    print(markdown)
```
